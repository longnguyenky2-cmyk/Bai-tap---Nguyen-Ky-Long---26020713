```python
from math import*
def So_hoan_hao(n:int):
    if n < 2:
        return False
    tong = 1
    for i in range(2,isqrt(n)+1):
        if n % i == 0:
            tong += i
            uoc_doi_xung = n // i
            if uoc_doi_xung != i:
                tong += uoc_doi_xung
    return tong == n 
assert So_hoan_hao(6) == True
assert So_hoan_hao(28) == True
assert So_hoan_hao(496) == True
assert So_hoan_hao(10) == False
assert So_hoan_hao(1) == False
```

```python
def fibonanci(n:int) -> int:
    if n < 0:
        return False
    fib1 = 1
    fib0 = 1
    i = 2
    while i <= n:
        fibn = fib0 + fib1
        fib0 = fib1
        fib1 = fibn
        i += 1
    return fib1
assert fibonanci(0) == 1
assert fibonanci(1) == 1
assert fibonanci(2) == 2
assert fibonanci(5) == 8
```
