```python
# ĐỀ BÀI : NHẬP d,m,y là ngày tháng năm của 1 ngày từ bàn phím, in ra ngày tiếp theo theo định dạng dd//mm/yy
# f"{d:02d}" là lệnh để thêm chữ số 0 nếu cần , ví dụ 1 thành 01 , 12 thành 12.
d,m,y = map(int, input().split())
A = [1,3,5,7,8,10,12]
B = [4,6,9,11]
if m in A:
    mx = 31
elif m in B:
    mx = 30
else:
    if y % 400 == 0 or ( y % 4 == 0 and y % 100 != 0):
        mx = 29
    else:
        mx = 28
# cap nhat ngay:
d += 1

if d > mx:
    d = 1
    m += 1
if y > 12:
    m = 1
    y += 1
print(f"{d:02d}"+'/'+f"{m:02d}"+'/'+str(y))
```
