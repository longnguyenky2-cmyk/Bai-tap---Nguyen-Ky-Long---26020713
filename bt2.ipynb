```python
# BAI 3 - HW
from math import*
a,b,c = map(float, input().split())
if a+b > c and b+c > a and a+c > b:
    if a == b == c:
        print("Tam giac deu")
    elif a == b != c or a == c != b or b == c != a:
        print('Tam giac can')
    elif a**2 + b**2 == c**2 or a**2 + c**2 == b**2 or c**2 + b**2 == a**2:
        print("Tam giac vuong")
    else:
        print("tam giac thuong")
    C = a+b+c
    p = C/2
    s = sqrt((p*(p-a)*(p-b)*(p-c)))
    print(C)
    print(s)

else:
    print("ko phai tam giac") 
```

```python
# BAI 5 - HW giai pt bac 2 ax^2 + bx + c = 0
from math import*
a,b,c = map(float, input().split())
delta = b**2 - 4*a*c
if delta > 0:
    can_delta = sqrt(delta)
    x1 = (-b-can_delta)/2*a
    x2 = (-b+can_delta)/2*a
    print(x1, x2)
elif delta == 0 :
    x1 = x2 = -b/2*a
    print(x1)
else:
    print('PT vo nghiem')
```

```python
# BAI 8:
a,b,c,d,m,n = map(float, input().split())
Da = a*d - b*c
Dx = m*d - b*n
Dy = a*n - m*c
if Da != 0:
    x = Dx / Da 
    y = Dy / Da 
    print(x,y) 
else:
    if Dx == 0 and Dy == 0:
        print('Vo so nghiem') 
    elif Dx != 0 or Dy != 0:
        print('He vo nghiem')
```

```python
# BAI 10:
from math import*
r = float(input())
a = float(input('Hoanh do cua tam la:'))
b = float(input('Tung do cua tam la:'))
x = float(input('Hoanh do cua diem la:'))
y = float(input('Tung do cua diem la:'))
khoang_cach = sqrt((x-a)**2+(y-b)**2)
if khoang_cach < r:
    print('Diem nam trong duong tron')
elif khoang_cach > r:
    print("Diem nam ngoai duong tron")
else:
    print('Diem nam tren duong tron')
```
