```python
from math import*
def So_hoan_hao(n:int):
    if n < 2:
        return False
    tong = 1
    for i in range(2,isqrt(n)+1):
        if n % i == 0:
            tong += i
            uoc_doi_xung = n // i
            if uoc_doi_xung != i:
                tong += uoc_doi_xung
    return tong == n 
assert So_hoan_hao(6) == True
assert So_hoan_hao(28) == True
assert So_hoan_hao(496) == True
assert So_hoan_hao(10) == False
assert So_hoan_hao(1) == False
        
   

```
